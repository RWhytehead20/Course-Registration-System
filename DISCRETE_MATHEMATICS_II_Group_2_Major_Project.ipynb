**Graph Representations**

```python
# Adjacency List Representation
class GraphListRepresentation:
    def __init__(self, total_nodes):
        self.total_nodes = total_nodes
        self.node_connections = {node: [] for node in range(total_nodes)}

    def insert_edge(self, source, target, weight=1):
        self.node_connections[source].append((target, weight))
        self.node_connections[target].append((source, weight))

    def show_graph(self):
        print("Graph (Adjacency List)")
        for node in self.node_connections:
            print(f" Node {node}: {self.node_connections[node]}")
```

```python
#Adjacency Matrix Representation
class GraphMatrixRepresentation:
    def __init__(self, total_nodes):
        self.total_nodes = total_nodes
        self.matrix_repr = [[0 for _ in range(total_nodes)] for _ in range(total_nodes)]

    def insert_edge(self, source, target, weight=1):
        self.matrix_repr[source][target] = weight
        self.matrix_repr[target][source] = weight

    def show_graph(self):
        print("Graph (Adjacency Matrix)")
        for row in self.matrix_repr:
            print(row)
```

```python
#Edge List Representation
class GraphEdgeListRepresentation:
    def __init__(self):

        self.edges_collection = []

    def insert_edge(self, source, target, weight=1):
        self.edges_collection.append((source, target, weight))

    def show_graph(self):
        print("Graph (Edge List)")
        for edge in self.edges_collection:
            print(f" Edge from {edge[0]} to {edge[1]} with weight {edge[2]}")
```

```python
#Incidence Matrix Representation
class GraphIncMatrixRepresentation:
    def __init__(self, total_nodes):
        self.total_nodes = total_nodes
        self.edge_pairs = []

    def insert_edge(self, source, target):
        self.edge_pairs.append((source, target))

    def show_graph(self):
        num_edges = len(self.edge_pairs)
        inc_matrix = [[0 for _ in range(num_edges)] for _ in range(self.total_nodes)]
        for idx, (source, target) in enumerate(self.edge_pairs):
            inc_matrix[source][idx] = 1
            inc_matrix[target][idx] = 1
        print("Graph (Incidence Matrix)")
        for row in inc_matrix:
            print(row)
```

```python
#Dictionary of Dictionaries Representation
class GraphDictRepresentation:
    def __init__(self):
        self.graph_structure = {}

    def insert_edge(self, source, target, weight=1):
        if source not in self.graph_structure:
            self.graph_structure[source] = {}
        if target not in self.graph_structure:
            self.graph_structure[target] = {}
        self.graph_structure[source][target] = weight
        self.graph_structure[target][source] = weight

    def show_graph(self):
        print("Graph (Dictionary of Dictionaries)")
        for node in self.graph_structure:
            print(f" Node {node}: {self.graph_structure[node]}")
```

```python
# Testing All Graph Representations

# Define sample edges as tuples (source, target, weight)
sample_edges = [
    (0, 1, 3),
    (0, 2, 5),
    (1, 2, 2),
    (1, 3, 4),
    (2, 4, 7)
]

total_nodes = 5

# Testing GraphListRepresentation (Adjacency List)
print("Testing Adjacency List Representation")
graph_list_obj = GraphListRepresentation(total_nodes)
for src, tgt, wt in sample_edges:
    graph_list_obj.insert_edge(src, tgt, wt)
graph_list_obj.show_graph()
print("\n")

# Testing GraphMatrixRepresentation (Adjacency Matrix)
print("Testing Adjacency Matrix Representation")
graph_matrix_obj = GraphMatrixRepresentation(total_nodes)
for src, tgt, wt in sample_edges:
    graph_matrix_obj.insert_edge(src, tgt, wt)
graph_matrix_obj.show_graph()
print("\n")

# Testing GraphEdgeListRepresentation (Edge List)
print("Testing Edge List Representation ")
graph_edge_list_obj = GraphEdgeListRepresentation()
for src, tgt, wt in sample_edges:
    graph_edge_list_obj.insert_edge(src, tgt, wt)
graph_edge_list_obj.show_graph()
print("\n")

# Testing GraphIncMatrixRepresentation (Incidence Matrix)
print("Testing Incidence Matrix Representation")
graph_inc_matrix_obj = GraphIncMatrixRepresentation(total_nodes)
# For incidence matrix, using unweighted edges.
for src, tgt, _ in sample_edges:
    graph_inc_matrix_obj.insert_edge(src, tgt)
graph_inc_matrix_obj.show_graph()
print("\n")

# Testing GraphDictRepresentation (Dictionary of Dictionaries)
print("Testing Dictionary of Dictionaries Representation")
graph_dict_obj = GraphDictRepresentation()
for src, tgt, wt in sample_edges:
    graph_dict_obj.insert_edge(src, tgt, wt)
graph_dict_obj.show_graph()
```

**Real World Application**

```python
import matplotlib.pyplot as plt
import networkx as nx
import heapq

# The graph is defined as a dictionary
# Each city is a key, and its value is a list of neigboring cities with distances
graph = {
    'A': [('B', 10), ('C', 15)],
    'B': [('A', 10), ('C', 5), ('D', 20)],
    'C': [('A', 15), ('B', 5), ('D', 10), ('E', 30)],
    'D': [('B', 20), ('C', 10), ('E', 5)],
    'E': [('D', 5), ('C', 30)]
}

# Dijksta's Algorithm Function. Finds shortest paths from start to all over vertices
def dijkstra(graph, start):
    distances = {vertex: float('inf') for vertex in graph} # All distances initialize to infinity
    distances[start] = 0 # Distance to start node is 0
    queue = [(0, start)] # Priority queue to hold (distance, vertex)
    previous = {vertex: None for vertex in graph} # Dict to track shortest path

    while queue:
      # Pop vertex with the smallest known distances
        current_distance, current_vertex = heapq.heappop(queue)

        # Explore neighbors of the current vertex
        for neighbor, weight in graph[current_vertex]:
            distance = current_distance + weight

            # If a shorter path is found
            if distance < distances[neighbor]:
              distances[neighbor] = distance
              previous[neighbor] = current_vertex
              heapq.heappush(queue, (distance, neighbor))

    return distances, previous

# Run the algorithm from startind node 'A'
start_node = 'A'
distances, previous = dijkstra(graph, start_node)

# Build and visualize the graph using NetworkX
G = nx.Graph()

# Add all edges to the graph
for node, edges in graph.items():
    for neighbor, weight in edges:
        G.add_edge(node, neighbor, weight=weight)

pos = nx.spring_layout(G, seed=42) # Spring layout for visual spacing
plt.figure(figsize=(10,6)) # Set up the plot
# Draw nodes and edges
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=2000, font_size=14, font_weight='bold')
# Label edges with weights (distances)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=12)

# Highlight shortest path edges in red
for target in distances:
    if target == start_node or distances[target] == float('inf'):
        continue

    # Reconstruct path from start to target
    path = []
    current = target
    while current:
        path.append(current)
        current = previous[current]
    path = path[::-1]
    # Create edge pairs
    path_edges = list(zip(path[:-1], path[1:]))
    nx.draw_networkx_edges(G, pos, edgelist=path_edges, width=3, edge_color='red')

# Annotate each node with its distance from the start node
for node in distances:
    x, y = pos[node]
    plt.text(x, y - 0.1, f"{distances[node]:.0f}", fontsize=10, color = 'black', ha='center')

# Final plot formatting
plt.title(f"Shortest paths from '{start_node}'", fontsize=16)
plt.axis('off')
plt.tight_layout()
plt.show()
```
